```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
hours = np.array([1,2,3,4,5]).reshape(-1,1)
marks = np.array([35,40,50,60,70])
```

```python
model = LinearRegression()
```

```python
model.fit(hours,marks)
```

```python
predicted = model.predict(hours)
```

```python
plt.scatter(hours, marks)   # Draws the actual data points
plt.plot(hours, predicted) # Draws the "Best Fit Line"
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Hours vs Marks (Linear Regression)")
plt.show()
```

```python
result = model.predict([[6]])
print("Predicted Marks:",result[0])
```
